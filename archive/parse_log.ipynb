```python
import pandas as pd
from log2csv import ips2geo, response2df, logfile2df
```

```python
# Path to the log file
log_file = '/media/phil/m2ssd/web/logs/www.philiplessner.com.access.log'
```

```python
df = logfile2df(log_file)
```

```python
df.head()
```

```python
# Filter for robots.txt requests with Google in user agent
mask = (df['request'].str.contains('robots.txt', case=False, na=False)) | \
        (df['user_agent'].str.contains('Google', case=False, na=False))

# Use .copy() to ensure downstream assignments won't raise SettingWithCopyWarning
df_filtered = df[mask].copy()
df = df[~mask].copy()

print(f"Filtered dataframe shape: {df_filtered.shape}")
print(f"Remaining dataframe shape: {df.shape}")
print(f"\nFiltered records:")
print(df_filtered)
print(f"\nRemaining records:")
print(df)
```
